```python
# 1. Basic Exploration
print("Dimensions of the dataset:", data.shape)
print("\nData types of each column:\n", data.dtypes)
print("\nSample rows from the dataset:\n", data.head())
```

```python
# 2. Handling Missing Values
missing_values = data.isnull().sum()
print("\nMissing values:\n", missing_values)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 3. Distribution of Categorical Variables
plt.figure(figsize=(10, 6))
sns.countplot(y='Job Title', data=data, order=data['Job Title'].value_counts().index[:10])
plt.title('Top 10 Job Titles')
plt.xlabel('Count')
plt.ylabel('Job Title')
plt.show()
```

```python

# Location Analysis
plt.figure(figsize=(10, 6))
top_locations = data['Location'].value_counts().head(10)
sns.barplot(x=top_locations.values, y=top_locations.index)
plt.title('Top 10 Locations with Highest Number of Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Location')
plt.show()

```

```python
# Location Analysis
plt.figure(figsize=(10, 6))
top_locations = data['Location'].value_counts().head(10)
sns.barplot(x=top_locations.values, y=top_locations.index)
plt.title('Top 10 Locations with Highest Number of Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Location')
plt.show()

```

```python
# Time-Series Analysis (Crawl Timestamp)
# Convert Crawl Timestamp to datetime format
data['Crawl Timestamp'] = pd.to_datetime(data['Crawl Timestamp'])

# Extract month and year from Crawl Timestamp
data['Crawl Month'] = data['Crawl Timestamp'].dt.month
data['Crawl Year'] = data['Crawl Timestamp'].dt.year

plt.figure(figsize=(10, 6))
sns.countplot(x='Crawl Month', hue='Crawl Year', data=data)
plt.title('Job Postings Over Time')
plt.xlabel('Month')
plt.ylabel('Number of Job Postings')
plt.legend(title='Year')
plt.show()
```
